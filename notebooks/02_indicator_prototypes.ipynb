#### **Connect to DuckDB & Query Summary Statistics**

```python
import duckdb
import pandas as pd

# Connect to DuckDB database
db_path = "../data/energy_transition.db"
conn = duckdb.connect(db_path)

# Test query: Check available tables and row counts
tables = conn.execute("SHOW TABLES;").fetchall()
print("Tables in DuckDB:", tables)

# Query latest available year's generation mix for Ethiopia as a test
query = """
    SELECT Clean_Country, Year, "Electricity source", "Generation (TWh)", "Share of generation (%)"
    FROM ember_data
    WHERE Clean_Country = 'Ethiopia' AND Year = 2024
    ORDER BY "Generation (TWh)" DESC;
"""

df_test = conn.execute(query).df()
print("\nTest Query Result (Ethiopia - 2024 Generation Mix):")
display(df_test.head(10))

# Close connection
conn.close()
```

#### **Prototype Core Technology Mix & Aggregate Filtering**

```python
import duckdb

conn = duckdb.connect("../data/energy_transition.db")

query_tech_breakdown = """
    SELECT 
        Clean_Country,
        Year,
        "Electricity source",
        "Generation (TWh)",
        "Share of generation (%)"
    FROM ember_data
    WHERE Year = 2024 
      AND "Is aggregated source" = 0  -- Filters for specific disaggregated technologies
    ORDER BY Clean_Country, "Generation (TWh)" DESC;
"""

df_tech = conn.execute(query_tech_breakdown).df()
print(f"Disaggregated Tech Records for 2024: {len(df_tech)}")
display(df_tech.head(15))

conn.close()
```
